```python
import json

items = sorted(json.load(open("config/items.txt")))
for item in items:
    print(f'"{item}",')
```

```python
from collections import defaultdict

known_colors = {
    "black", "blue", "brown", "cyan", "gray", "green", "light_blue", "light_gray*", "lime", "magenta", "orange", "pink", "purple", "red", "white", "yellow"
}
groups = defaultdict(list)

for item in items:
    if "suppsquared" not in item:
        continue
    for pattern in known_colors:
        if pattern in item:
            prefix = item.replace(pattern, '')
            groups[prefix].append(item)

subsets = [v for v in groups.values() if len(v) > 1]

for sub in subsets:
    for item in sub:
        print(f'"{item}",')
    print("")
```

```python
patterns = {"_log", "_leaves", "_sapling", "package"}
groups = defaultdict(list)

for item in items:
    for pattern in patterns:
        if pattern in item.split(":")[1]:
            groups[pattern].append(item)

subsets = [v for v in groups.values() if len(v) > 1]

for sub,key in zip(subsets, groups.keys()):
    print(key)
    for item in sub:
        print(f'"{item}",')
    print("")
```

```python
import networkx as nx
import matplotlib.pyplot as plt

g = nx.DiGraph()
g.add_nodes_from((1,2,3,4))
g.add_edge(1,2,weight=3)
g.add_edge(2,3,weight=5)
g.add_edge(3,4,weight=7)
g.add_edge(1,4,weight=5)
g.add_edge(4,2,weight=6)

nx.draw_spring(g, with_labels=True)
plt.show()

tag = "23"
nodes = {2,3}

for i,o,data in g.edges(data=True):
    print(i,o,data)

for inc,out,data in g.copy().edges().data():
    if inc in nodes:
        g.add_edge(tag, out, **data)
    if out in nodes:
        g.add_edge(inc, tag, **data)

for node in g.copy().nodes():
    if node in nodes:
        g.remove_node(node)

nx.draw_spring(g, with_labels=True)
plt.show()

for i,o,data in g.edges(data=True):
    print(i,o,data)
```

```python
import json
recipes = json.load(open("config/recipes.json"))
for r in recipes.keys():
    for ingr in list(recipes[r]["input"].keys()):
        # Renames the ingredients
        new = ingr.replace("net.minecraft.world.item.crafting.", "")
        recipes[r]["input"][new] = recipes[r]["input"].pop(ingr)

        # Removes empty ingredients (like for shaped recipes)
        if new == "Ingredient@1":
            recipes[r]["input"].pop(new)


for r in recipes.keys():
    if "protection_pixel" in r and len(recipes[r]["input"]) == 0:
        print(r, list(recipes[r]["output"].keys())[0])
        print()
```
