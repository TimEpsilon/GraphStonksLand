```python
from core.GraphCreator import GraphCreator
from core.PropagationAlgorithm import Propagation
```

```python
gc = GraphCreator("items.txt", "recipes.json")
```

```python
prop = Propagation(gc.G)
prop.generateAtomicInputs()
```

```python
from ipywidgets import BoundedFloatText, Select, VBox, HBox, Output
from IPython.display import display

rows = []

def make_row(index, row):
    widgets_in_row = []

    if row["cycle"] is None:
        float_widget = BoundedFloatText(
            value=row["value"],
            min=0,
            max=10_000,
            step=0.001,
            description=f"{row['node']} -> ",
            disabled=False
        )

        def on_float_change(change):
            prop.inputs.at[index, "value"] = change["new"]

        float_widget.observe(on_float_change, names='value')
        widgets_in_row.append(float_widget)

    else:
        select_widget = Select(
            options=row["cycle"],
            value=list(row["cycle"])[0],
            description=f'{row["node"]}',
            disabled=False
        )
        float_widget = BoundedFloatText(
            value=row["value"],
            min=0,
            max=10_000,
            step=0.001,
            description=f" -> ",
            disabled=False
        )

        def on_select_change(change):
            prop.inputs.at[index, "representative"] = change["new"]

        def on_float_change(change):
            prop.inputs.at[index, "value"] = change["new"]

        select_widget.observe(on_select_change, names='value')
        float_widget.observe(on_float_change, names='value')

        widgets_in_row.extend([select_widget, float_widget])

    return HBox(widgets_in_row)

for idx, row in prop.inputs.iterrows():
    rows.append(make_row(idx, row))

display(VBox(rows))
```

```python
prop.saveAtomicInputs()
```

```python
prop.loadAtomicInputs()
prop.setupForSolving()
prop.solve()
```

```python
output = prop.getOutputs()
print(output.to_markdown())
output.to_json("SCT_Table.json")
```
