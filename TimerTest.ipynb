```python
stats = {}
uuid = {}

for i,f in enumerate(glob('D:/Documents/Stonks/stonkslandV2_last/54b5f102-00aa-4f22-87d1-164bc662867e/stonksland2/stats/**')):
    name = f.split("\\")[-1].replace(".json", "")
    stats[i] = pd.read_json(f).T
    uuid[i] = name
print(stats[1]["minecraft:custom"]["stats"])
```

```python
play = []
death = []
for i in stats:
    play.append(stats[i]["minecraft:custom"]["stats"]["minecraft:play_time"]/60/60/20)
    death.append(stats[i]["minecraft:custom"]["stats"]["minecraft:deaths"])
    print(names[uuid[i]], play[-1], death[-1])

plt.figure()
plt.hist(play,bins=18)
plt.axvline(np.median(play), color='r')
plt.axvline(np.mean(play), color='b')
plt.show()
print(np.median(play), np.mean(play))

plt.figure()
plt.hist(death,bins=18)
plt.axvline(np.median(death), color='r')
plt.axvline(np.mean(death), color='b')
plt.show()
print(np.median(death), np.mean(death))
```
